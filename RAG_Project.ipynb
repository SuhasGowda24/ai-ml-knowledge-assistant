## RAGAS Evaluation

The RAG pipeline was evaluated using 10 question-and-ground-truth-answer pairs.
Faithfulness measures whether generated answers are supported by the retrieved
context, while answer relevancy measures how relevant the answers are to the
questions.

The following averages were calculated from the saved per-question evaluation
results.

```python
import json
import statistics
import pandas as pd

with open(
    "evaluation_results_before_final_evaluation.json",
    "r",
    encoding="utf-8"
) as file:
    evaluation_results = json.load(file)

faithfulness_scores = [
    item["faithfulness"]
    for item in evaluation_results
    if item.get("faithfulness") is not None
]

relevancy_scores = [
    item["answer_relevancy"]
    for item in evaluation_results
    if item.get("answer_relevancy") is not None
]

evaluation_summary = pd.DataFrame({
    "Metric": ["Faithfulness", "Answer Relevancy"],
    "Average Score": [
        round(statistics.mean(faithfulness_scores), 4),
        round(statistics.mean(relevancy_scores), 4)
    ],
    "Evaluated Questions": [
        len(faithfulness_scores),
        len(relevancy_scores)
    ]
})

display(evaluation_summary)
```

## Vector Store Persistence and Reproducibility

The project uses ChromaDB as its vector database to store document embeddings and retrieve relevant passages for user queries.

**Vector Store Configuration**

- **Vector database:** ChromaDB
- **Collection name:** `ai_ml_knowledge`
- **Embedding model:** `sentence-transformers/all-MiniLM-L6-v2`
- **Indexed document chunks:** 242
- **Persistence directory:** `vectorstore/`

**Rebuilding the Vector Store**

To reproduce the vector store from the original research PDFs:

1. Place the source PDF documents in the `documents/` directory.
2. Install the dependencies listed in `requirements.txt`.
3. Run the document loading and chunking pipeline.
4. Generate embeddings and create the ChromaDB index.
5. Save the index to the `vectorstore/` directory.
6. Load the persisted index when running the RAG pipeline.

Run the following command from the project directory to build the vector store:

`python build_vectorstore.py`

The source PDFs, project scripts, dependencies, and write permissions for the persistence directory are required to rebuild the index. The notebook's existing outputs demonstrate the previously executed pipeline; they do not independently provide all the files required for a fresh rebuild.
